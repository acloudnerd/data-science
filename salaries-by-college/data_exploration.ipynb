```python
# find college major with highest starting salaries 

clean_df['Starting Median Salary'].idxmax()
```

```python
# clean_df['Undergraduate Major'].loc[43]
clean_df['Undergraduate Major'][43]
```

```python
# College with the highest Mid Career Median Salary

pos = clean_df['Mid-Career Median Salary'].idxmax()
the_major = clean_df['Undergraduate Major'][pos]
max_salary = clean_df['Mid-Career Median Salary'].max()

print(f"{the_major} has the highest median salary which amounts to: {max_salary}")
```

```python
min_pos = clean_df['Starting Median Salary'].idxmin()
the_min_major = clean_df['Undergraduate Major'][min_pos]
min_salary = clean_df['Starting Median Salary'].min()

print(f"{the_min_major} has the minimum starting salary which amounts to: {min_salary}")
```

```python
low_mid_salary_pos = clean_df["Mid-Career Median Salary"].idxmin()
the_mid_low_major = clean_df['Undergraduate Major'][low_mid_salary_pos]
low_mid_salary_amt = clean_df['Mid-Career Median Salary'].min()

print(f"The college major that has the lowest mid-career salary is {the_mid_low_major} and the amount is: {low_mid_salary_amt}")
```

```python
# diff between mid career 90th and 10th percentile salary

spread_mid_10th_90th = clean_df['Mid-Career 90th Percentile Salary'] - clean_df['Mid-Career 10th Percentile Salary']
```

```python
spread_mid_10th_90th.head()
```

```python
clean_df.insert(6, 'spread_mid_10th_90th', spread_mid_10th_90th)
```

```python
sorted_data_by_spread = clean_df.sort_values('spread_mid_10th_90th', ascending=True)
sorted_data_by_spread[['Undergraduate Major', 'spread_mid_10th_90th']].head()
```
